```python
# Assignment 4

# First matrix
A = [
    [1, 2, 3],
    [4, 5, 6]
]

# Second matrix
B = [
    [7, 8, 9],
    [10, 11, 12]
]

# Matrix to store the result
C = [
    [0, 0, 0],
    [0, 0, 0]
]

# Addition of two matrices
for i in range(2):
    for j in range(3):
        C[i][j] = A[i][j] + B[i][j]

# Display matrices
print("First Matrix:")
for row in A:
    print(row)

print("\nSecond Matrix:")
for row in B:
    print(row)

print("\nSum of the Matrices:")
for row in C:
    print(row)
```
