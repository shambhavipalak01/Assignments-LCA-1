```python
# Assignment 1

# Dictionary to store student records
students = {
    101: ("Shambhavi", "CSE", 90),
    102: ("Arshita", "ECE", 85),
    103: ("Rajvi", "AIML", 78)
}

# Display initial records
print("Initial Student Records:")
for roll, details in students.items():
    print("Roll No:", roll, "| Name:", details[0],
          "| Branch:", details[1], "| Marks:", details[2])

# 1. Add a new student record
students[104] = ("Kanak", "IT", 88)
print("\nAfter Adding New Student:")
print(students)

# 2. Delete an existing student record
del students[102]
print("\nAfter Deleting Student with Roll No. 102:")
print(students)

# 3. Update the details of a student
students[103] = ("Rajvi", "CSE", 92)
print("\nAfter Updating Student with Roll No. 103:")
print(students)

# 4. Display final student records
print("\nFinal Student Records:")

# Converting dictionary items into a list
student_list = list(students.items())

for roll, details in student_list:
    print("Roll No:", roll,
          "| Name:", details[0],
          "| Branch:", details[1],
          "| Marks:", details[2])
```
